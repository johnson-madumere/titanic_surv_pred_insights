### 1. Determine Business Objectives


#### 1.1 Background

The Titanic disaster resulted in a large loss of life, but survival was not evenly distributed among passengers. Historical evidence suggests that personal and socio-economic factors such as gender, passenger class, age, and family relationships influenced survival chances.

This project uses historical Titanic passenger data to analyze survival patterns and build prediction tables that explain who was more likely to survive and why.


#### 1.2 Business Objectives  
The main objectives are:  
- Build a machine learning model that predicts whether a passenger survived the Titanic disaster.  
- Identify and explain the key factors (such as gender, passenger class, age, and family size) that influenced survival chances.  
- Generate structured, interpretable outputs to support analysis and insight for both technical and non-technical audiences.

To achieve this, the project will generate the following outputs:  
- **(A) Core Prediction Insight**: Compares actual survival with model predictions.  
- **(B) Enriched Prediction Insight**: Combines predictions with key human-readable passenger features.  
- **(C) Group-Level Insight**: Shows survival rates by passenger group.  
- **(D) Feature Impact Insight**: Explains which factors influenced survival the most.  
- **(E) Smart Prediction & Insight**: Summarizes survival likelihood and reasons.  

***Focus:*** Not only prediction but also understanding survival behavior and patterns.  

#### 1.3 Business Success Criteria  
The project is considered successful if:  
- A predictive model is built that can reasonably distinguish between survivors and non-survivors.  
- Survival probabilities are generated for each passenger, not just binary outcomes.  
- Patterns in survival (e.g., by gender, class, age) are clearly identified.  
- Tables and charts are structured, readable, and interpretable for non-technical audiences.
