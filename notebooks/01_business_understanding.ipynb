#### 1.1 Background

The Titanic disaster caused a large loss of life. However, not all passengers had the same chance of surviving. Historical records suggest that several factors influenced survival. These include gender, passenger class, age, and whether someone was traveling with family.

In this project, I use historical Titanic passenger data [2],[3] to explore survival patterns. The goal is to understand which factors were linked to survival and to build prediction tables that estimate the likelihood of survival for different passengers.

#### 1.2 Vision

**FOR** safety researchers, data science learners, students, and decision-support analysts  
**WHO** want clear evidence about survival patterns, model confidence, and fairness across passenger groups 
**THE** Titanic Survival Insight System is an interpretable prediction and reporting system  
**THAT** estimates survival probability, highlights the main factors behind predictions, and provides reports on confidence, fairness, and prediction errors  
**UNLIKE** black-box prediction workflows that mainly focus on overall accuracy  
**THIS PRODUCT** provides a structured and easy-to-read analytics approach that combines prediction results, explanations, confidence measures, and fairness analysis.

#### 1.3 Business Objectives  
The main objectives of this project are::  
- Build a machine learning model that predicts whether a Titanic passenger survived (1) or did not survive (0).  
- Produce prediction outputs that are clear and understandable for both technical and non-technical audiences.  
- Explain model confidence, fairness patterns, and prediction errors so that the results can be interpreted responsibly. 

To support these objectives, the project will generate the following outputs in the final phase:  
- **(A) Prediction Certainty and Confidence Report:** Shows the probability distribution of predictions and checks how reliable they are.

- **(D) Core Prediction Insight Table:** Displays actual survival outcomes, predicted survival, and the main reason behind each prediction.
- **(E) Enriched Prediction Insight Table:** Combines passenger details with prediction results and probabilities.
- **(F) Smart Output Insight Table:** Provides a simple summary of survival likelihood and the key factor influencing the prediction.

The main ***focus:*** of this project is not only prediction accuracy, but also clear explanations, model confidence, and useful insights[2], [3].

#### 2.4 Terminology  

To keep the notebooks consistent, the following terms are used throughout the project:

- **Prediction Certainty and Confidence Report:** - Shows how confident the model is in its predictions based on probability scores. 
- **Group Fairness and Accuracy Analysis:** - Compares model performance across groups such as gender or passenger class. 
- **Error Breakdown Analysis:** - Shows where the model makes mistakes, including false positives, false negatives, and misclassified passengers.  
- **Core Prediction Table:** - Displays actual survival results, predicted outcomes, and the main reason for the prediction.
- **Enriched Prediction Table:** - Combines passenger information with prediction results and probability values.
- **Smart Output Table:** - Provides a short and simple summary of survival likelihood and the key reason behind the prediction. 
- **Survival Probability:** - The predicted probability that a passenger survived. 
- **Main Reason:** - The most important feature (primary driver) that influenced the prediction.
- **CRISP-DM:** - Cross-Industry Standard Process for Data Mining
- **Survival Prediction** - The task of predicting whether a Titanic passenger survived (1) or did not survive (0).
- **Target Variable** - The variable the model aims to predict. In this project, the target variable is Survived
- **Feature (Predictor Variable)** - An input variable used by the model to make predictions.
- **Feature Engineering** – The process of creating new variables from existing data to make patterns clearer.
- **Pclass_Code** - A numeric representation of passenger class, where 1 represents First Class, 2 Second Class, and 3 Third Class.
- **SibSp** - This represents the number of siblings or spouses traveling with the passenger.
- **Parch** - This represents the number of parents or children traveling with the passenger.
- **Data Cleaning** - The process of correcting errors and inconsistencies in the dataset.
- **Missing Value Imputation** - The process of replacing missing data with estimated values.
- **Exploratory Data Analysis (EDA)** - The process of analyzing and visualizing data to understand patterns, relationships, and potential issues before modeling.
- **Class Imbalance** – A situation where one outcome occurs more often than the other. In the Titanic dataset, more passengers did not survive than survived.
- **Stratified Sampling** – A data splitting method that keeps the same proportion of survival outcomes across the training, validation, and test datasets.
- **Training Set** – The portion of the dataset used to train machine learning models
- **Validation Set** - A dataset used to tune model parameters and compare model performance.
- **Test Set** – A dataset used to evaluate the final performance of the model after training and tuning.
- **Data Leakage** – A problem that occurs when information related to the target variable is accidentally included in the model features, which can lead to overly optimistic results.
- **Model-Ready Dataset** – A cleaned and formatted dataset where missing values are handled, categorical variables are encoded, and all features are prepared for machine learning models.
- **Normalization** – A data preprocessing technique that rescales numerical features so they fall within a similar range, often between 0 and 1.
- **Feature Scaling** – A general term for methods used to adjust the range of numerical variables so that no single feature dominates the model due to its larger numeric values.
- **Standardization** – A type of feature scaling where values are transformed so the feature has a mean of 0 and a standard deviation of 1.
- **Regularization** – A technique used during model training to prevent overfitting by adding a penalty for large model coefficients. It helps the model generalize better to unseen data.
- **Overfitting** – A situation where a model learns the training data too closely, including noise or random patterns, which causes poor performance on new data.
- **L1 Regularization (Lasso)** – A regularization method that adds a penalty based on the absolute values of model coefficients.
- **L2 Regularization (Ridge)** – A regularization method that adds a penalty based on the squared values of model coefficients.
- **Hyperparameter** – A configuration setting chosen before training a model, such as the regularization strength in logistic regression. Hyperparameters influence how the model learns from data.

#### 4.4 Acceptance criteria
The models should generate probability estimates and meet predefined performance thresholds.

**For example:**
- Accuracy ≥ predefined threshold
- AUC ≥ predefined threshold
- If these thresholds are not reached, the results must be clearly explained and justified.

**Quantitative Acceptance Thresholds**

A model is considered acceptable if, on the held-out test set, it meets:

- **AUC-ROC >= 0.78** (primary ranking quality)
- **Log Loss <= 0.58** (probability quality)
- **Accuracy >= 0.77**
- **Recall (Survived=1) >= 0.70** (avoid missing survivors)
- **Precision (Survived=1) >= 0.72**
- **Calibration check:** reliability curve deviation is acceptable (no systematic overconfidence)
- **Fairness check:** subgroup recall gap across Sex and Pclass should be <= 0.15 where sample size is adequate

If one threshold fails but business interpretation value is strong, the exception must be documented with justification and improvement plan.

#### 4.6 Model Selection Rationale & Trade-offs

For this project, I use **three candidate models** in the modeling phase:

- **Logistic Regression**
- **Decision Tree**
- **Random Forest**

These models were chosen because they balance prediction performance with interpretability. [4]

**Why?**
- **Logistic Regression** Provides a simple and interpretable baseline model for binary classification.
- **Decision Tree** Captures decision rules and non-linear relationships between variables.
- **Random Forest** Combines multiple decision trees to improve prediction performance.


**Trade-offs:**
- **Interpretability vs. predictive strength**  
  Logistic Regression and Decision Trees are easier to explain. Random Forest may perform better but is less transparent.[4]

- **Simple patterns vs. complex patterns**  
  Logistic Regression works well when relationships are linear. Tree-based models can capture more complex interactions.[4]

- **Overfitting control**  
  Logistic Regression uses regularization. Tree models control overfitting using parameters such as *max_depth* and *min_samples_leaf*. [4]

**Decision:**  
Instead of selecting one model early, all three models will be trained using the same dataset split. Their performance and interpretability will be compared in the evaluation phase.

#### 4.7 Validation & Evaluation Plan

- **Data split strategy:**
  - The dataset will be divided using a hold-out approach [4]:
    - Training set: 70%
    - Validation set: 15%
    - Test set: 15%
- **Evaluation metrics:**
  - Primary: AUC-ROC or Log Loss (since probabilities required) [4]
  - Secondary: Accuracy, Precision, Recall, F1, and confusion matrix [4]
- **Performance thresholds & statistical tests:**
  - Define minimal acceptable values (e.g., AUC ≥ 0.75) or justify alternatives.
  - Use bootstrap or paired tests to compare models and check significance.
- **Planned visualizations:**
  - ROC curves, calibration plots, feature coefficient plots, partial dependence or SHAP summary plots.
